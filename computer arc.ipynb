```python
def logical_left_shift(binary, positions):
    print("\nLogical Left Shift")
    print("Original 8-bit number:", binary)

    result = binary

    for i in range(positions):
        result = result[1:] + '0'
        print("After shift", i + 1, ":", result)

    print("Final 8-bit result:", result)


def logical_right_shift(binary, positions):
    print("\nLogical Right Shift")
    print("Original 8-bit number:", binary)

    result = binary

    for i in range(positions):
        result = '0' + result[:-1]
        print("After shift", i + 1, ":", result)

    print("Final 8-bit result:", result)


# (a) Logical left shift
num1 = "11110110"
logical_left_shift(num1, 2)

# (b) Logical right shift
num2 = "10010100"
logical_right_shift(num2, 2)
```

```python
def to_twos_complement(number, bits):
    """Convert a signed integer to its two's complement representation."""
    return format(number & ((1 << bits) - 1), f'0{bits}b')


def signed_binary_add(a, b):
    # Display operands in 8-bit two's complement
    a8 = to_twos_complement(a, 8)
    b8 = to_twos_complement(b, 8)

    # Perform addition using 16-bit two's complement
    a16 = to_twos_complement(a, 16)
    b16 = to_twos_complement(b, 16)

    result = a + b
    result16 = to_twos_complement(result, 16)
    result8 = to_twos_complement(result, 8)

    # Check signed 8-bit overflow
    overflow = result < -128 or result > 127

    print("\nOperand A (8-bit):", a8)
    print("Operand B (8-bit):", b8)
    print("Operand A (16-bit):", a16)
    print("Operand B (16-bit):", b16)

    print("\n16-bit Binary Addition:")
    print("  ", a16)
    print("+ ", b16)
    print("-" * 22)
    print("  ", result16)

    print("\nResult (16-bit binary):", result16)
    print("Result (decimal):", result)
    print("Result (8-bit binary):", result8)
    print("8-bit Arithmetic Overflow:", "Yes" if overflow else "No")


# Expression (a)
print("Expression (a): 24 + (+17)")
signed_binary_add(24, 17)

# Expression (b)
print("\nExpression (b): (-25) + (-18)")
signed_binary_add(-25, -18)
```
